```python
def check_line_props(linefit):
    #print(linefit['cen_err'],linefit)

    good=False
    
    notgood=False
    for key in linefit:
        if linefit[key]==None:
            notgood=True
            return False
            #print('bad key')
    if linefit['cen_err']>1.0 or linefit['wid_err']>1.0:
        print('in check line props ',linefit['cen_err']>1.0, linefit)
        return False
    if notgood==False and linefit['amp']<0 and linefit['r']>0.995 and linefit['wid']>0 and (linefit['wid_err']/linefit['wid'])<.3 and (linefit['amp_err']/linefit['amp'])<.3:
        
        good=True

        if linefit['cen_err']>1.0: 
            print(linefit)
            print('error in check line props ',good)
    else:
        good=False
    return good
```

```python
def flatten(xss):
    return [x for xs in xss for x in xs]
```

```python
goodlines=[]

for l in lines:
    goodfits_NEID=[]
    total_fits_NEID=0
    goodfits_EXPRES=[]
    total_fits_EXPRES=0
    for fit in data:
        for linefit in fit['NEID_fits']:
            if linefit['line_wl']==l:
                total_fits_NEID+=1
                good=check_line_props(linefit)
                if good:
                    goodfits_NEID.append(linefit)
                    #print('good fit!')
        for linefit in fit['EXPRES_fits']:
            if linefit['line_wl']==l:
                total_fits_EXPRES+=1
                good=check_line_props(linefit)
                if good:
                    goodfits_EXPRES.append(linefit)
                    #print('good fit!')
    num_good_fits_NEID=len(goodfits_NEID)
    num_good_fits_EXPRES=len(goodfits_EXPRES)

    if  num_good_fits_EXPRES>30 and num_good_fits_NEID>30:
        goodlines.append(l)
```

```python
line_time_series_NEID={}
line_time_series_EXPRES={}
dates=[]
for lnum,l in enumerate(goodlines[:]):
    print(lnum,l)
    line_time_series_EXPRES[str(lnum)]={}
    line_time_series_NEID[str(lnum)]={}
    for key in (data[0])['NEID_fits'][0].keys():
        line_time_series_EXPRES[str(lnum)][key]=[]
        line_time_series_NEID[str(lnum)][key]=[]

    #for each window, see if there are good measurements for the line with both EXPRES and NEID
    NEID_orders=[]
    EXPRES_orders=[]

    for fit in data:
        NEID_orders_window=[]
        EXPRES_orders_window=[]
        #for each fit, get all the orders with this line
        for linefit in fit['NEID_fits']:
            #if linefit is fot this line
            if linefit['line_wl']==l:
                good=check_line_props(linefit)
                if good:
                    NEID_orders_window.append(linefit['order'])
        for linefit in fit['EXPRES_fits']:
            #if linefit is fot this line
            if linefit['line_wl']==l:
                good=check_line_props(linefit)
                if good:
                    EXPRES_orders_window.append(linefit['order'])

        NEID_orders.append(NEID_orders_window)
        EXPRES_orders.append(EXPRES_orders_window)

    unique_NEID_orders=list(set(flatten(NEID_orders)))
    unique_EXPRES_orders=list(set(flatten(EXPRES_orders)))

    if len(unique_NEID_orders)>1:
        #pick the better order
        counts=[flatten(NEID_orders).count(item) for item in unique_NEID_orders]

        #for now, the better order is the first order if they have the same number of good fits
        best_NEID_order=unique_NEID_orders[np.argmax(counts)]
    else:
        best_NEID_order=unique_NEID_orders[0]

    if len(unique_EXPRES_orders)>1:
        #pick the better order
        counts=[flatten(EXPRES_orders).count(item) for item in unique_EXPRES_orders]

        #for now, the better order is the first order if they have the same number of good fits
        best_EXPRES_order=unique_EXPRES_orders[np.argmax(counts)]
    else:
        best_EXPRES_order=unique_EXPRES_orders[0]        

    for fit in data[:]:
        NEID_good=False
        EXPRES_good=False
        if lnum==1:
            dates.append(fit['overlap_mid'])
        #for each fit, get all the orders with this line
        for linefit_N_t in fit['NEID_fits']:
            if linefit_N_t['line_wl']==l and linefit_N_t['order']==best_NEID_order:
                #linefit matches
                NEID_good=check_line_props(linefit_N_t)
                linefit_N=linefit_N_t
                continue
        for linefit_E_t in fit['EXPRES_fits']:
            if linefit_E_t['line_wl']==l and linefit_E_t['order']==best_EXPRES_order:
                #linefit matches
                EXPRES_good=check_line_props(linefit_E_t) 
                linefit_E=linefit_E_t
                key='cen_err'
                if linefit_E[key]!=None:
                    #print(linefit_E[key])
                    if linefit_E[key]>100:
                        print('bad cen err found at ',fit['index'], str(lnum))
                        print(linefit_E)
                        print(EXPRES_good)

                continue

        if EXPRES_good and NEID_good:
            for key in (data[0])['NEID_fits'][0].keys():
                line_time_series_EXPRES[str(lnum)][key].append(linefit_E[key])
                line_time_series_NEID[str(lnum)][key].append(linefit_N[key])

        else:
            for key in (data[0])['NEID_fits'][0].keys():
                line_time_series_EXPRES[str(lnum)][key].append(np.nan)
                line_time_series_NEID[str(lnum)][key].append(np.nan)

        
```

```python
correlations={}

keys_to_use=['wid','cen','amp']
for key in keys_to_use:
    correlations[key]=[]
    for lnum in line_time_series_NEID.keys():
        N=np.array(line_time_series_NEID[str(lnum)][key])
        E=np.array(line_time_series_EXPRES[str(lnum)][key])
        good_dates=np.where(~np.isnan(N))
        correlations[key].append(np.corrcoef(N[good_dates],E[good_dates])[0][1])
```
